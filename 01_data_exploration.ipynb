```python
import pandas as pd

# ── STEP 1: Define columns to load from raw files ──────────
cols_2025 = [
    'YEAR_MONTH', 'PRACTICE_CODE',
    'BNF_CODE',  # will rename to BNF_PRESENTATION_CODE
    'ITEMS', 'TOTAL_QUANTITY', 'NIC', 'ACTUAL_COST', 'ADQUSAGE'
]

cols_2026 = [
    'YEAR_MONTH', 'PRACTICE_CODE',
    'BNF_PRESENTATION_CODE',
    'ITEMS', 'TOTAL_QUANTITY', 'NIC', 'ACTUAL_COST', 'ADQ_USAGE'
]

# ── STEP 2: Load full files ─────────────────────────────────
print("Loading 2025 file...")
df_2025 = pd.read_csv("/lakehouse/default/Files/epd_snomed_202501.csv",
                       usecols=cols_2025,
                       dtype={'YEAR_MONTH': str, 'PRACTICE_CODE': str, 'BNF_CODE': str})

df_2025 = df_2025.rename(columns={
    'BNF_CODE': 'BNF_PRESENTATION_CODE',
    'ADQUSAGE': 'ADQ_USAGE'
})
df_2025['source_year'] = 2025
print("2025 loaded:", df_2025.shape)

print("Loading 2026 file...")
df_2026 = pd.read_csv("/lakehouse/default/Files/epd_snomed_202601.csv",
                       usecols=cols_2026,
                       dtype={'YEAR_MONTH': str, 'PRACTICE_CODE': str, 'BNF_PRESENTATION_CODE': str})
df_2026['source_year'] = 2026
print("2026 loaded:", df_2026.shape)

# ── STEP 3: Combine both years ──────────────────────────────
df_raw = pd.concat([df_2025, df_2026], ignore_index=True)
print("Combined raw rows:", df_raw.shape)

# ── STEP 4: Load dimension tables for key lookup ───────────
print("Loading dimensions...")
dim_geography = pd.read_parquet("/lakehouse/default/Files/dim_geography.parquet")[['geography_key', 'PRACTICE_CODE']]
dim_drug = pd.read_parquet("/lakehouse/default/Files/dim_drug.parquet")[['drug_key', 'BNF_PRESENTATION_CODE']]
dim_date = pd.read_parquet("/lakehouse/default/Files/dim_date.parquet")[['date_key', 'YEAR_MONTH']]

# ── STEP 5: Join dimension keys ─────────────────────────────
print("Joining geography keys...")
df_fact = df_raw.merge(dim_geography, on='PRACTICE_CODE', how='left')

print("Joining drug keys...")
df_fact = df_fact.merge(dim_drug, on='BNF_PRESENTATION_CODE', how='left')

print("Joining date keys...")
df_fact = df_fact.merge(dim_date, on='YEAR_MONTH', how='left')

# ── STEP 6: Keep only keys and measures ────────────────────
fact_columns = ['date_key', 'geography_key', 'drug_key',
                'ITEMS', 'TOTAL_QUANTITY', 'NIC', 'ACTUAL_COST', 'ADQ_USAGE']

fact_prescriptions = df_fact[fact_columns]

# ── STEP 7: Check for nulls in keys ────────────────────────
print("\nNull check on keys:")
print(fact_prescriptions[['date_key', 'geography_key', 'drug_key']].isnull().sum())

# ── STEP 8: Save to Files section ──────────────────────────
print("\nSaving fact table...")
fact_prescriptions.to_parquet("/lakehouse/default/Files/fact_prescriptions.parquet", index=False)

print("\nFact table saved:", fact_prescriptions.shape)
print("Done")
```

```python
import pandas as pd

# ── STEP 1: Define columns to load ─────────────────────────
cols_2025 = [
    'YEAR_MONTH', 'PRACTICE_CODE',
    'BNF_CODE',
    'ITEMS', 'TOTAL_QUANTITY', 'NIC', 'ACTUAL_COST', 'ADQUSAGE'
]

cols_2026 = [
    'YEAR_MONTH', 'PRACTICE_CODE',
    'BNF_PRESENTATION_CODE',
    'ITEMS', 'TOTAL_QUANTITY', 'NIC', 'ACTUAL_COST', 'ADQ_USAGE'
]

# ── STEP 2: Load full files ─────────────────────────────────
print("Loading 2025 file...")
df_2025 = pd.read_csv("/lakehouse/default/Files/epd_snomed_202501.csv",
                       usecols=cols_2025,
                       dtype={'YEAR_MONTH': str, 'PRACTICE_CODE': str, 'BNF_CODE': str})

df_2025 = df_2025.rename(columns={
    'BNF_CODE': 'BNF_PRESENTATION_CODE',
    'ADQUSAGE': 'ADQ_USAGE'
})
df_2025['source_year'] = 2025
print("2025 loaded:", df_2025.shape)

print("Loading 2026 file...")
df_2026 = pd.read_csv("/lakehouse/default/Files/epd_snomed_202601.csv",
                       usecols=cols_2026,
                       dtype={'YEAR_MONTH': str, 'PRACTICE_CODE': str, 
                              'BNF_PRESENTATION_CODE': str})
df_2026['source_year'] = 2026
print("2026 loaded:", df_2026.shape)

# ── STEP 3: Combine both years ──────────────────────────────
df_raw = pd.concat([df_2025, df_2026], ignore_index=True)
print("Combined raw rows:", df_raw.shape)

# ── STEP 4: Fix YEAR_MONTH format ──────────────────────────
# Raw data has '2026-01' but dim_date has '202601' — remove dash
df_raw['YEAR_MONTH'] = df_raw['YEAR_MONTH'].str.replace('-', '', regex=False)
print("YEAR_MONTH sample after fix:", df_raw['YEAR_MONTH'].head(3).tolist())

# ── STEP 5: Load dimension keys only ───────────────────────
print("Loading dimensions...")
dim_geography = pd.read_parquet("/lakehouse/default/Files/dim_geography.parquet")[['geography_key', 'PRACTICE_CODE']]
dim_drug = pd.read_parquet("/lakehouse/default/Files/dim_drug.parquet")[['drug_key', 'BNF_PRESENTATION_CODE']]
dim_date = pd.read_parquet("/lakehouse/default/Files/dim_date.parquet")[['date_key', 'YEAR_MONTH']]

# ── STEP 6: Join dimension keys ─────────────────────────────
print("Joining geography keys...")
df_fact = df_raw.merge(dim_geography, on='PRACTICE_CODE', how='left')

print("Joining drug keys...")
df_fact = df_fact.merge(dim_drug, on='BNF_PRESENTATION_CODE', how='left')

print("Joining date keys...")
df_fact = df_fact.merge(dim_date, on='YEAR_MONTH', how='left')

# ── STEP 7: Keep only keys and measures ────────────────────
fact_columns = ['date_key', 'geography_key', 'drug_key',
                'ITEMS', 'TOTAL_QUANTITY', 'NIC', 'ACTUAL_COST', 'ADQ_USAGE']

fact_prescriptions = df_fact[fact_columns]

# ── STEP 8: Null check ──────────────────────────────────────
print("\nNull check on keys:")
print(fact_prescriptions[['date_key', 'geography_key', 'drug_key']].isnull().sum())

# ── STEP 9: Save to Files section ──────────────────────────
print("\nSaving fact table...")
fact_prescriptions.to_parquet("/lakehouse/default/Files/fact_prescriptions.parquet", index=False)

print("\nFact table saved:", fact_prescriptions.shape)
print("Done")
```
